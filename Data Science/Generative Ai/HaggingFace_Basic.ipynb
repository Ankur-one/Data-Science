```python
from transformers import pipeline
```

```python
# create a classfier pipeline
classifier = pipeline('sentiment-analysis')
```

```python
# Classifiy the text
text = 'I am Happy but i am unhealthy'
result = classifier(text)
print(result)
```

# Text Generation

```python
# Take a generator
generator = pipeline(task="text-generation")
```

```python
# Generate a text using generator
text = generator("The world is a", max_length=50, do_simple-False)
print(text[0]['generator_text'])
```
