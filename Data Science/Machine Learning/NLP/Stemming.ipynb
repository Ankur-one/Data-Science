# RegrexStemmer class

## RegexpStemmer is a stemming class in NLTK that removes a specified pattern from words using a regular expression (regex).

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer("ing$")

words = ["playing", "running", "walking", "eating", "play"]

for word in words:
    print(word, "---->", stemmer.stem(word))
```

# Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")
stemmer.stem("studies")
```
